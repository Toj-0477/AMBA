# BigMart Sales Prediction - Final predictions

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge

train = pd.read_csv('train_clean.csv')
test = pd.read_csv('test_clean.csv')
sample = pd.read_csv('sample_submission.csv')

print('train:', train.shape)
print('test: ', test.shape)
print('sample submission:', sample.shape)
```

## Ridge Regression


```python
columns = ['Item_Fat_Content', 'Item_Type', 'Outlet_Identifier', 'Outlet_Size',
           'Outlet_Location_Type', 'Outlet_Type', 'Item_Category',
           'Base_Price', 'Item_MRP', 'Item_Weight', 'Item_Visibility',
           'Outlet_Age', 'Price_by_weight']

all_rows = pd.get_dummies(pd.concat([train[columns], test[columns]]))

X_train = all_rows[:len(train)]
X_test = all_rows[len(train):]
y_train = train['Item_Outlet_Sales']
```

```python
model = Ridge(alpha=44.0)
model.fit(X_train, y_train)

final_prediction = model.predict(X_test)
print(final_prediction[:5].round(2))
```

## Save predictions


```python
# sales can't be negative
final_prediction = np.clip(final_prediction, 0, None)

submission = pd.DataFrame()
submission['Item_Identifier'] = test['Item_Identifier']
submission['Outlet_Identifier'] = test['Outlet_Identifier']
submission['Item_Outlet_Sales'] = final_prediction

submission.to_csv('submission.csv', index=False)
submission.head()
```
